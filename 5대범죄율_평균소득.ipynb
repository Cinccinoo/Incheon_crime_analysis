```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import statsmodels.api as sm

# 파일 경로 설정
crime_file = '/content/crime_123.csv'
income_file = '/content/평균소득 보정값.xlsx'

# 데이터 불러오기
crime_df = pd.read_csv(crime_file)  # CSV 파일 불러오기
income_df = pd.read_excel(income_file)  # 엑셀 파일 불러오기

# 데이터 확인
print(crime_df.head())
print(income_df.head())
```

```python
# 필요한 라이브러리 불러오기
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# 데이터: 평균소득 보정값을 사용하여 외국인 비율 대신 평균소득으로 변경
data_income = {
    '군구': ['연수구', '계양구', '서구', '부평구', '미추홀구', '남동구', '중구', '강화군'],
    '평균소득': [0.99, 1.17, 0.56, 0.61, 0.71, 0.55, 1.88, 3.01]
}
df_income = pd.DataFrame(data_income)

# 범죄율 데이터
data_crime_rate = {
    '군구': ['중구', '미추홀구', '남동구', '부평구', '서구', '계양구', '강화군', '연수구'],
    '살인': [0.006962, 0.017179, 0.015435, 0.002054, 0.009002, 0.010145, 0.028697, 0.012832],
    '강도': [0.020886, 0.026996, 0.025083, 0.028761, 0.005401, 0.023672, 0.014348, 0.012832],
    '강간.강제추행': [0.772802, 0.620913, 0.329942, 0.201329, 0.428535, 0.378767, 0.258275, 0.318239],
    '절도': [6.084952, 4.0568, 1.724962, 2.237219, 2.337138, 2.492424, 2.439269, 3.169559],
    '폭력': [9.405916, 5.639762, 2.656906, 2.652203, 4.893946, 4.372734, 4.835492, 3.695681]
}
df_crime_rate = pd.DataFrame(data_crime_rate)

# 데이터 합치기 (군구를 기준으로 병합)
merged_df_income = pd.merge(df_income, df_crime_rate, on='군구')

# 데이터 준비
crime_columns = ['강도', '강간.강제추행', '절도', '폭력']
income_column = '평균소득'

# 상관관계 히트맵
plt.figure(figsize=(10, 8))
sns.heatmap(merged_df_income[crime_columns + [income_column]].corr(), annot=True, cmap='coolwarm', fmt=".2f")
plt.title("5대범죄율과 평균소득 상관관계")
plt.show()

# 산점도 시각화: 평균소득 vs 범죄 유형
fig, axes = plt.subplots(nrows=3, ncols=2, figsize=(12, 15))
axes = axes.flatten()

for i, column in enumerate(crime_columns):
    sns.scatterplot(data=merged_df_income, x=income_column, y=column, ax=axes[i])
    axes[i].set_title(f"{income_column} vs {column}")
    axes[i].set_xlabel("평균소득")
    axes[i].set_ylabel(column)

# 여분의 subplot 제거
for j in range(len(crime_columns), len(axes)):
    fig.delaxes(axes[j])

plt.tight_layout()
plt.show()
```
